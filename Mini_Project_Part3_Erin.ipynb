Q9.Find species–island groups with at least 20 penguins.

### **Part 3 — Combining DataFrames + GroupBy**

Q12. Create the following penguin_regions dataframe and merge with penguins.



```
{
    'island': ['Biscoe', 'Dream', 'Torgersen'],
    'region': ['Antarctica', 'Antarctica', 'Antarctica']
}
```



````python
regions_dict = {
    'island': ['Biscoe', 'Dream', 'Torgersen'],
    'region': ['Antarctica', 'Antarctica', 'Antarctica']
}
regions = pd.DataFrame(regions_dict)
penguins_x_regions = penguins.join(regions.set_index('island'), on='island',how='outer')
penguins_x_regions
````
