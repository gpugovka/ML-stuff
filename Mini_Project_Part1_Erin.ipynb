### **STEP 3 Handle Missing Values:**

*Identify missing values in the dataset*

```python
# Identify missing values in the dataset.

def show_missing_stats(peng):
# Count number of notnull values per column, number of missing values per column and percentage of missing values per column
  result = pd.concat([peng.notnull().sum(), peng.isna().sum(), round(peng.isna().mean() * 100, 2)],
              axis=1,
              keys=["valid_count", "missing_count", "missing_percent"])
  return result

show_missing_stats(penguins)
```

*Clearly explain your chosen strategy for handling missing data (e.g., imputation, dropping rows/columns) and justify your decision.*

I'm seeing two "classes" of missing data.
1. Four fields have a small portion, just 0.58%, missing. For this class droping the rows with missing data will result in at most 0.58% * 4 ~= 2% total data loss for all of them together, which is a minor loss.

Let's first handle these.

```python
# Drop rows where measurements are missing
penguins = penguins.dropna(subset=["body_mass_g", "flipper_length_mm", "bill_depth_mm", "bill_length_mm"]).copy()
show_missing_stats(penguins)
```

2. The field "sex" has total of 2.91% missing data which is more than all other missing data together. Dropping it will cause a relatively meaningful data loss, so we should consider opportunities for making an affort of reasonable approximation for the missing data in this column.

One approach that can be used for the column "sex" in meaning of biological sex is trying to predict it based on the known biological features of the penguin,for example based on the weight of the penguin.



```python
def species_median_mass_by_sex_per_species(peng):
  # Calculates median body mass per sex
  species_median_mass = peng.dropna(subset=["sex","species"]).groupby(["sex","species"])["body_mass_g"].median()
  return species_median_mass

species_median_mass_by_sex_per_species(penguins)
```

On our way to non-empty data we discover another error in the dataset: there's "." entered for sex somewhere. Let's treat it as a missing data.

```python
penguins["sex"] = penguins["sex"].replace(".", np.nan)

median_mass_by_sex_per_species = species_median_mass_by_sex_per_species(penguins)
median_mass_by_sex_per_species
```

It's clean of the "." now, let's continue.
Calculate midpoint threshold between female and male medians. Let's consider everything above it in weight a male penguin, and below a female

```python
# Calculate mass "thresholds" for each species as average mass value and later
# split female vs male approximation using it
thresholds = median_mass_by_sex_per_species.groupby(level="species").mean()
thresholds
```

```python

# For each species compare the threshold average mass value with the mass of the
# penguin. If it is lighter - consider a female, otherwise a male.
penguins["sex"] = penguins["sex"].fillna(
    (penguins["body_mass_g"] >= penguins["species"].map(thresholds)).map({True: "MALE", False: "FEMALE"})
)
penguins
```

```python
# Check for missing data again

show_missing_stats(penguins)
```

```python
# Verify if mean has changed (should not)
pd.concat([penguins.mean(numeric_only=True), initial_mean], axis=1)
```

```python
# Verify if std has changed (should not)
pd.concat([penguins.std(numeric_only=True), initial_std], axis=1)
```

The stats are unchanged, which is expacted for the approach we used.

### **STEP 4 Perform EDA and Visualization**

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
species_list = penguins["species"].unique()
colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]

for sp, col in zip(species_list, colors):
    subset = penguins[penguins["species"] == sp]["flipper_length_mm"]
    ax.hist(subset, bins=15, alpha=0.6, label=sp, color=col, edgecolor="black")

ax.set_title("Flipper Length Distribution by Species")
ax.set_xlabel("Flipper Length (mm)")
ax.set_ylabel("Count")
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

Justification: Evaluates spread, modality, and skewness of continuous variables across groups.Insights: Highlights a distinct bimodal shape overall. Gentoo flippers range higher 210 to 230 mm, while Adelie and Chinstrap heavily overlap in the lower band 180 to 200 mm.

```python
fig, ax = plt.subplots(figsize=(8, 4))
data_by_species = [
    penguins[penguins["species"] == sp]["body_mass_g"]
    for sp in species_list
]

ax.boxplot(data_by_species, tick_labels=species_list, patch_artist=True)
ax.set_title("Body Mass Across Species")
ax.set_xlabel("Species")
ax.set_ylabel("Body Mass (g)")
ax.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

Justification: Shows five-number summaries (median, IQR, whiskers) and flags outliers without distributional assumptions.Insights: Gentoo median weight 5000 g is significantly higher than Adelie and Chinstrap 3700 g, showing distinct body mass boundaries with minimal overlap.

```python
fig, ax = plt.subplots(figsize=(8, 5))

for sp, col in zip(species_list, colors):
    subset = penguins[penguins["species"] == sp]
    ax.scatter(subset["bill_length_mm"], subset["bill_depth_mm"], label=sp, color=col, alpha=0.8, edgecolors="none")

ax.set_title("Bill Length vs. Bill Depth by Species")
ax.set_xlabel("Bill Length (mm)")
ax.set_ylabel("Bill Depth (mm)")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.7)
plt.show()
```

Justification: Maps two continuous dimensions against each other to identify clusters and correlations.

Insights: Shows a negative correlation across species splits into positive linear trends within each group. Adelie has short/deep bills, Chinstrap has long/deep bills, and Gentoo has long/shallow bills.
